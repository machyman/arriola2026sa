**Purpose:** Implement the adjoint method for linear systems. Verify Theorem 5.1 (adjoint = forward SA). Illustrate the LP duality connection and cost comparison.

**Key claims tested:**
- Adjoint gives identical SI as direct solve (Theorem 5.1); cost is r solves vs m solves; LP duality: A^T v = c is dual constraint

```python
import numpy as np
import matplotlib.pyplot as plt
import warnings; warnings.filterwarnings('ignore')
FULL = False
print('Chapter 5: The Adjoint in Linear Algebra')
```

```python
# ── Forward vs Adjoint for linear system Au = b ───────────────────────────────
# Example from §5.1: A = [[3,1],[1,2]], b depends on parameters q1, q2
# Response: J = u1 + u2 = c^T u, c = [1,1]

A = np.array([[3.0, 1.0], [1.0, 2.0]])
c = np.array([1.0, 1.0])

# Forward (FSE): one solve per parameter
b = lambda q: np.array(q)
q_nom = np.array([1.0, 1.0])
u_nom = np.linalg.solve(A, b(q_nom))
J_nom = c @ u_nom

db_dq1 = np.array([1.0, 0.0])  # db/dq1
db_dq2 = np.array([0.0, 1.0])  # db/dq2
du_dq1 = np.linalg.solve(A, db_dq1)  # FSE solve 1
du_dq2 = np.linalg.solve(A, db_dq2)  # FSE solve 2
dJ_dq1_fwd = c @ du_dq1
dJ_dq2_fwd = c @ du_dq2

# Adjoint: ONE solve regardless of number of parameters
v = np.linalg.solve(A.T, c)          # A^T v = c (one solve)
dJ_dq1_adj = v @ db_dq1              # dot product only
dJ_dq2_adj = v @ db_dq2              # dot product only

print('Forward SA (2 solves):')
print(f'  dJ/dq1 = {dJ_dq1_fwd:.6f}  dJ/dq2 = {dJ_dq2_fwd:.6f}')
print('Adjoint SA (1 solve + 2 dot products):')
print(f'  dJ/dq1 = {dJ_dq1_adj:.6f}  dJ/dq2 = {dJ_dq2_adj:.6f}')
err = max(abs(dJ_dq1_fwd-dJ_dq1_adj), abs(dJ_dq2_fwd-dJ_dq2_adj))
assert err < 1e-12, f'FAIL: err={err}'
print(f'PASS: Forward = Adjoint (Theorem 5.1), max diff = {err:.2e}')
print(f'Adjoint vector v = {v}  (encodes weighted contributions to J)')
```

```python
# ── Verification: Theorem 5.1, the adjoint identity, and the cost claim ──────
# Self-contained apart from variables defined in the cell above.
# Every quantity is checked against a closed form, not against itself.
import numpy as np

_tol = 1e-12

# (1) Closed form. det(A) = 5, so A^{-1} = (1/5)[[2,-1],[-1,3]] and, with
#     c = b(q_nom) = [1,1], the exact sensitivities are 1/5 and 2/5.
_Ainv = np.array([[2.0, -1.0], [-1.0, 3.0]]) / 5.0
assert np.allclose(_Ainv @ A, np.eye(2), atol=_tol), "closed-form inverse is wrong"
_exact = np.array([1.0/5.0, 2.0/5.0])

# (2) Theorem 5.1: the two routes agree, and both agree with the closed form.
_fwd = np.array([dJ_dq1_fwd, dJ_dq2_fwd])
_adj = np.array([dJ_dq1_adj, dJ_dq2_adj])
assert np.max(np.abs(_fwd - _adj)) < _tol, (
    f"forward and adjoint disagree by {np.max(np.abs(_fwd-_adj)):.3e}")
assert np.max(np.abs(_adj - _exact)) < _tol, (
    f"adjoint route gives {_adj}, closed form gives {_exact}")
assert abs(J_nom - 3.0/5.0) < _tol, f"J_nom = {J_nom}, expected 3/5"

# (3) The adjoint variable satisfies its defining equation A^T v = c.
assert np.max(np.abs(A.T @ v - c)) < _tol, "A^T v = c is not satisfied"

# (4) The cost claim of the chapter: forward needs one solve per parameter,
#     adjoint needs one solve per response, independent of m.
_m, _r = 2, 1                      # 2 parameters (q1, q2), 1 response J
_forward_solves, _adjoint_solves = _m, _r
assert _adjoint_solves < _forward_solves, (
    "with m > r the adjoint must use strictly fewer solves")

print(f"Verification passed: dJ/dq = {_adj} = exact [1/5, 2/5]; "
      f"A^T v = c to {np.max(np.abs(A.T @ v - c)):.1e}; "
      f"{_adjoint_solves} adjoint solve vs {_forward_solves} forward.")
```

```python
# ── Cost comparison figure ────────────────────────────────────────────────────
m_vals = np.arange(1, 51)
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(m_vals, m_vals, 'c-', lw=2, label='Forward SA: m linear solves')
for r, col, ls in [(1,'steelblue','-'),(3,'coral','--'),(5,'seagreen',':')]:
    ax.axhline(r, color=col, lw=2, ls=ls, label=f'Adjoint: r={r} QOI(s)')
ax.set_xlabel('Number of POIs (m)', fontsize=12)
ax.set_ylabel('Number of linear solves', fontsize=12)
ax.set_title('Adjoint vs Forward SA — Cost Comparison\n(Theorem 5.1: same answer, lower cost)', fontsize=11)
ax.legend(fontsize=9); ax.grid(alpha=0.3)
ax.set_xlim(1,50); ax.set_ylim(0,51)
plt.tight_layout()
plt.savefig('ch05_fig_cost.pdf', dpi=150, bbox_inches='tight')
plt.show(); print('Fig 5 saved.')
```

```python
# ── Verification Suite, §5 Verification table ────────────────────────────────
# The cell above checks Theorem 5.1 on the §5.1 linear
# system.  The table in §5 Verification uses the competitive Lotka-Volterra
# equilibrium, and before this cell no notebook computed it.  That table is the
# chapter's central claim, that the adjoint is not an approximation, so it is
# the one a reader most needs to be able to run.
import numpy as np

print("=" * 66)
print("VERIFICATION SUITE  —  Chapter 5, the FSE/adjoint table")
print("=" * 66)

_a, _b, _c = 0.4, 1.0, 0.3           # nominal values stated in §5 Verification
_den = 1.0 - _a*_c
_u = np.array([(1.0 - _a)/_den, (1.0 - _c)/_den])

# Model Jacobian of f_1 = u1(1-u1-a u2), f_2 = u2(1-u2-c u1) at equilibrium.
_J = np.array([[1 - 2*_u[0] - _a*_u[1], -_a*_u[0]],
               [-_c*_u[1],              1 - 2*_u[1] - _c*_u[0]]])

# -df/dp for each parameter.  b enters neither equation, hence the zero column.
_rhs = {'a': np.array([_u[0]*_u[1], 0.0]),
        'b': np.array([0.0, 0.0]),
        'c': np.array([0.0, _u[0]*_u[1]])}

# Route 1, FSE: one linear solve per parameter, three solves.
_fse = {p: np.linalg.solve(_J, r) for p, r in _rhs.items()}

# Route 2, adjoint: one solve per QOI, two solves.  For QOI_i = u_i the adjoint
# variable is lambda_i = J^{-T} e_i and the sensitivity is lambda_i . (-df/dp).
_lam = [np.linalg.solve(_J.T, e) for e in np.eye(2)]
_adj = {p: np.array([_lam[i] @ _rhs[p] for i in range(2)]) for p in _rhs}

# Normalized sensitivity indices, S^{u_i}_p = (p/u_i) du_i/dp.
_pv = {'a': _a, 'b': _b, 'c': _c}
def _norm(d):
    return {p: np.array([_pv[p]/_u[i] * d[p][i] for i in range(2)]) for p in d}
_S_fse, _S_adj = _norm(_fse), _norm(_adj)

# Test 1 — the two routes agree, which is the claim the table is making.
_gap = max(np.max(np.abs(_S_fse[p] - _S_adj[p])) for p in _pv)
assert _gap < 1e-12, f"FAIL: FSE and adjoint differ by {_gap:.2e}; Theorem 5.1 requires 0"
print(f"Test 1 PASS  FSE (3 solves) and adjoint (2 solves) agree to {_gap:.1e}")

# Test 2 — both rows match the printed table.
_text = {('a', 0): -0.530, ('b', 0): 0.0, ('c', 0): 0.136,
         ('a', 1): +0.136, ('b', 1): 0.0, ('c', 1): -0.292}
for (p, i), want in _text.items():
    for route, S in (("FSE", _S_fse), ("adjoint", _S_adj)):
        got = S[p][i]
        assert abs(got - want) < 5e-4, \
            f"FAIL: {route} S^u{i+1}_{p} = {got:+.4f}, table says {want:+.3f}"
print("Test 2 PASS  both rows match the printed table to 3 dp:")
for route, S in (("FSE (3 solves)    ", _S_fse), ("Adjoint (2 solves)", _S_adj)):
    print(f"   {route}  " + "  ".join(f"{S[p][i]:+.3f}" for i in (0, 1) for p in ('a','b','c')))

print("=" * 66)
# Test 3 — the printed digits, one tagged check per printed row, so each
# number is tied to the row that prints it (Test 2 compares through _text,
# which no tag can reach).  The b column prints 0, an integer, and is left out.
assert np.allclose([_S_fse['a'][0], _S_fse['c'][0], _S_fse['a'][1], _S_fse['c'][1]],
                   [-0.530, 0.136, +0.136, -0.292], rtol=0, atol=5e-4), 'FAIL: FSE row'  # book: ch05_body.tex "FSE (3 solves)"
assert np.allclose([_S_adj['a'][0], _S_adj['c'][0], _S_adj['a'][1], _S_adj['c'][1]],
                   [-0.530, 0.136, +0.136, -0.292], rtol=0, atol=5e-4), 'FAIL: adjoint row'  # book: ch05_body.tex "Adjoint (2 solves)"
print("Test 3 PASS  printed digits checked row by row")
print("Chapter 5 table verification: 3 of 3 PASS")
print("=" * 66)
```

```python
# ── Verification Suite: Appendix A exercises ─────────────────────────────────
# f(t) = e^{-t}, g(t) = cos(pi t) on [0, 1].  Appendix A, Exercises, self-check answer.
import numpy as np
from scipy.integrate import quad

def agrees(x, shown, ndigits):
    """True when x rounds to the value the book prints, at the book's precision."""
    return abs(x - shown) <= 0.5 * 10.0 ** (-ndigits) + 1e-12

ip_fg = quad(lambda t: np.exp(-t) * np.cos(np.pi * t), 0, 1)[0]
nf2 = quad(lambda t: np.exp(-2 * t), 0, 1)[0]
ng2 = quad(lambda t: np.cos(np.pi * t) ** 2, 0, 1)[0]
assert abs(ip_fg - (np.exp(-1) + 1) / (1 + np.pi**2)) < 10**-12, "FAIL: closed form"
assert agrees(ip_fg, 0.126, 3), f"FAIL: <f,g> = {ip_fg:.6f}"  # book: appendix_a.tex "(1+\pi^2) \approx 0.126" "Cauchy--Schwarz: $|0.126|"
assert agrees(nf2, 0.432, 3), f"FAIL: |f|^2 = {nf2:.6f}"  # book: appendix_a.tex "(1-e^{-2})/2 \approx 0.432"
assert agrees(np.sqrt(ng2), 0.707, 3), "FAIL: |g|"  # book: appendix_a.tex "1/\sqrt{2} \approx 0.707" "\times 0.707 \approx"
# |f| = sqrt(0.432332) = 0.65752, which rounds to 0.658; drafts before v1_22_1 printed 0.657,
# and the Cauchy-Schwarz product 0.464 inherited it (exact 0.465).  The inequality holds either way.
assert round(1000 * np.sqrt(nf2)) == 658 and round(1000 * np.sqrt(nf2 * ng2)) == 465, "D-A4-1 values changed"
assert abs(ip_fg) <= np.sqrt(nf2 * ng2), "FAIL: Cauchy-Schwarz"
print(f"Appendix A PASS: <f,g> = {ip_fg:.4f}, |f| = {np.sqrt(nf2):.5f}, |g| = {np.sqrt(ng2):.4f}")
# Appendix A's adjoint and ANOVA self-checks and Appendix B's adjoint self-check, computed, not read off.
# Appendix A: L = d/dt on [0, 1], u = t^2, v = t; L* v = -dv/dt.
u_a, v_a, _h = (lambda t: t**2), (lambda t: t), 1e-6
Lu = lambda t: (u_a(t + _h) - u_a(t - _h)) / (2 * _h)
ip_Lu_v = quad(lambda t: Lu(t) * v_a(t), 0, 1)[0]
ip_u_Lsv = quad(lambda t: u_a(t) * -(v_a(t + _h) - v_a(t - _h)) / (2 * _h), 0, 1)[0]
bdry = u_a(1) * v_a(1) - u_a(0) * v_a(0)
assert agrees(Lu(0.7) / 0.7, 2, 6) and agrees(ip_Lu_v, 2/3, 6), 'FAIL: Lu = 2t and <Lu, v> = 2/3'  # book: appendix_a.tex "(a) $Lu = 2t$"
assert agrees(ip_Lu_v, ip_u_Lsv + bdry, 6) and agrees(ip_Lu_v, 2/3, 6), 'FAIL: the boundary identity'  # book: appendix_a.tex "\Rightarrow 2/3 = -1/3 + 1 = 2/3"
# Appendix A: f = Q1 + Q2 + Q1 Q2 with Q_i ~ U(0, 1); f0 = E[f] by double quadrature.
from scipy.integrate import dblquad
Ef = dblquad(lambda q2, q1: q1 + q2 + q1 * q2, 0, 1, 0, 1)[0]
assert agrees(Ef, 5/4, 9), f'FAIL: E[f] = {Ef}'  # book: appendix_a.tex "= 1/2 + 1/2 + 1/4 = 5/4$"
# Appendix B: A = [[3, 1], [0, 2]] (its transpose is printed), J = c^T u with c = (1, 0).
A_b, c_b = np.array([[3.0, 1.0], [0.0, 2.0]]), np.array([1.0, 0.0])
v_b = np.linalg.solve(A_b.T, c_b)
_Jb = lambda b: c_b @ np.linalg.solve(A_b, b)
_b0, _e1 = np.array([1.0, 1.0]), np.array([1e-6, 0.0])
dJ_db1 = (_Jb(_b0 + _e1) - _Jb(_b0 - _e1)) / 2e-6
assert agrees(dJ_db1, 2/6, 9) and agrees(v_b[0], 2/6, 12), f'FAIL: dJ/db1 = {dJ_db1}'  # book: appendix_b.tex "(c) $\partial J/\partial b_1 = 2/6 = 1/3 = v_1$"
```
